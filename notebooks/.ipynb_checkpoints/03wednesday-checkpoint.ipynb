# Wednesday: Volumetric Analysis

## 1. Behavioral Analysis (Volumetric Testing)
### Stress-Testing against High-Volume Floods
* **Focus:** DoS Hulk and DDoS signatures.
* **Metric:** Documenting the "Accuracy Paradox" (64% score with 0.00 recall).

```python
import pandas as pd
import numpy as np
import joblib
import os

if os.path.exists('tuesday_anomaly_detector.pkl'):
    model_tue = joblib.load('tuesday_anomaly_detector.pkl')
    print("✅ Tuesday's model is ready for testing on Wednesday data!")
```

## 2. Retraining with the "Clean" Data
### Large-Scale File Processing
* **Action:** Strategic sampling to prevent memory crashes on local hardware.
* **Discovery:** Universal Model requirement established.

```python
df_wed = pd.read_csv('Wednesday-workingHours.pcap_ISCX.csv', nrows=100000)

df_wed.columns = df_wed.columns.str.strip()
df_wed['Label'] = df_wed['Label'].str.strip()

df_wed.replace([np.inf, -np.inf], np.nan, inplace=True)
df_wed.dropna(inplace=True)
```

```python
strict_drop_list = ['Label', 'IsAnomaly', 'Flow ID', 'Source IP', 'Destination IP', 
                    'Timestamp', 'Source Port', 'Destination Port']

X_wed = df_wed.drop(columns=[c for c in strict_drop_list if c in df_wed.columns])
```

```python
wed_preds = model_tue.predict(X_wed)

y_wed = df_wed['Label'].apply(lambda x: 0 if x == 'BENIGN' else 1)

from sklearn.metrics import classification_report
print("Transfer Test: Tuesday Model predicting Wednesday Data")
print(classification_report(y_wed, wed_preds))
```
