```python
import os
import shutil

# 1. Define the folders
folders = ['data', 'models', 'notebooks', 'output']

# 2. Create folders if they don't exist
for folder in folders:
    if not os.path.exists(folder):
        os.makedirs(folder)

# 3. Move the files
for filename in os.listdir('.'):
    # Move CSVs to data/
    if filename.endswith('.csv'):
        shutil.move(filename, os.path.join('data', filename))
        print(f"Moved {filename} to data/")
        
    # Move Models to models/
    elif filename.endswith('.pkl'):
        shutil.move(filename, os.path.join('models', filename))
        print(f"Moved {filename} to models/")
        
    # Move Notebooks to notebooks/
    elif filename.endswith('.ipynb') and filename != 'Organizer.ipynb':
        shutil.move(filename, os.path.join('notebooks', filename))
        print(f"Moved {filename} to notebooks/")

print("\n✨ Workspace Clean!")
```
