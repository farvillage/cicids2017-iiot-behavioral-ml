## 2. Retraining with the "Clean" Data
### Dataset-Specific Sanitization
* **Action:** Managing Tuesday-specific "Infinity" values and header noise.
* **Insight:** Brute Force patterns do not overlap with Web Attack signatures.

```python
import os
import joblib
import pandas as pd

# Paths using the '../' navigation
model_path = '../models/thursday_model.pkl'
data_path = '../data/Tuesday-WorkingHours.pcap_ISCX.csv'

if os.path.exists(model_path):
    model_strict = joblib.load(model_path)
    print("Thursday Model loaded from /models!")
else:
    print(f"Error: {model_path} not found.")

if os.path.exists(data_path):
    df = pd.read_csv(data_path)
    print("Tuesday data loaded from /data!")
```

```python
df.columns = df.columns.str.strip()

df['Label'] = df['Label'].str.strip()

df.replace([np.inf, -np.inf], np.nan, inplace=True)
df.dropna(inplace=True)

strict_drop_list = ['Label', 'IsAnomaly', 'Flow ID', 'Source IP', 'Destination IP', 
                    'Timestamp', 'Source Port', 'Destination Port']

X_tuesday = df.drop(columns=[c for c in strict_drop_list if c in df.columns])
```

```python
tuesday_preds = model_strict.predict(X_tuesday)

y_tuesday = df['Label'].apply(lambda x: 0 if x == 'BENIGN' else 1)

print(classification_report(y_tuesday, tuesday_preds))
```
